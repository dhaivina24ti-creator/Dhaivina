```python
import pandas as pd

df = pd.read_excel('/content/drive/MyDrive/SEMESTER 5/Probstat/dhaivina.xlsx')

print("Data sudah berhasil dibaca.")
print("Total baris dan kolom: ", df.shape)
```

```python
# Tabel Frekuensi
datafrq = pd.crosstab(
    index=df["Grade"],
    columns="Frekuensi"
)

print(datafrq)
```

```python
datafrq.plot(color='magenta')

plt.show()
```

```python
datafrq.plot(
    kind='bar',
    color='#B39DDB',
    title='Bar Chart Frekuensi Grade'
)

plt.ylabel("Frekuensi")
plt.xlabel("Grade")
plt.show()
```

```python
# Pie Chart
datafrq.plot(
    kind='pie',
    y='Frekuensi',
    autopct='%1.0f%%',
    legend=False,
    colors=['#F8BBD0', '#A5D6A7', '#F48FB1', '#81C784', '#FCE4EC', '#C8E6C9']
)

plt.ylabel("")
plt.title("Pie Chart Frekuensi Grade")
plt.show()
```

```python
nilai = df["Final Score"]

tabel_statistik = pd.DataFrame({

    "Statistik": [
        "Mean", "Median", "Modus", "Minimum", "Maximum",
        "Q1", "Q2", "Q3", "Variance",
        "Standard Deviation", "Skewness", "Kurtosis", "Standard Error"
    ],

    "Nilai": [
        nilai.mean(),
        nilai.median(),
        nilai.mode().tolist(),
        nilai.min(),
        nilai.max(),
        nilai.quantile(0.25),
        nilai.quantile(0.50),
        nilai.quantile(0.75),
        nilai.var(),
        nilai.std(),
        nilai.skew(),
        nilai.kurt(),
        nilai.sem()
    ]

})

tabel_statistik.style.set_table_styles([
    {"selector": "th", "props": [
        ("background-color", "#B39DDB"),
        ("color", "white")
    ]},
    {"selector": "td", "props": [
        ("background-color", "#F3E5F5")
    ]}
])
```
